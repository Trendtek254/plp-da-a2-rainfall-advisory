# County Rainfall And Planting Advisory


**Dataset**

```python
import numpy as np
counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

clean_rows = [
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],   # Nairobi
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],    # Nakuru
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],   # Kisumu
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70]    # Mombasa
]

machakos_raw = ["48", "35", "88", "N/A", "140", " 20 ",
                "9", "12", "", "45", "182", "96"]
months = ["January","February","March","April","May",
          "June","July","August","September","October","November","December"]
```

**Task 1:**
***Our Data Cleaning Function***

```python
def machakos_clean(value):
  try:
    if value.strip() == "":
      return None
    else:
      return int(value)
  except ValueError:
    return None
cleaned_values =[]
for value in machakos_raw:
  cleaned_values.append(machakos_clean(value))

#Extracting valid values
valid_values =[value for value in cleaned_values if value is not None]

#Calculating the average
average = np.mean(valid_values)
average_rounded = round(average)


#Replacing missing values with the average
final_values = []
for value in cleaned_values:
  if value is None:
    final_values.append(average_rounded)
  else:
    final_values.append(value)

#Identifying the imputed months
imputed_months = []
for i in range(len(cleaned_values)):
  if cleaned_values[i] is None:
    imputed_months.append(months[i])




```

**Task 2**

```python
import numpy as np
all_counties = np.array(clean_rows + [final_values])
print(all_counties.dtype)
print(all_counties.shape)
print(all_counties)
```

**Task 3**

```python
annual_total_rain = np.sum(all_counties, axis=1)
average_rainfall = np.average(all_counties, axis=0)
print(annual_total_rain)
print(average_rainfall)

```

**Task 4**

```python
wettest_months = np.argmax(all_counties, axis=1)
wettest_month_names = [months[i] for i in wettest_months]

print(wettest_month_names)
```

**Task 5**

```python
dry_months = all_counties < 50

counts = np.sum(dry_months, axis=1)
for county,count in zip(counties,counts):
  print(f"{county} : {count} dry months")
```

**Task 6**

```python
first_months = all_counties[:,:-1]
second_months = all_counties[:,1:]
combined_rainfalls = first_months + second_months
highest_combinations = np.argmax(combined_rainfalls, axis=1)

# Convert months to a numpy array to allow indexing with an array
months_arr = np.array(months)
first_month_names = months_arr[highest_combinations]
second_month_names = months_arr[highest_combinations + 1]
planting_windows = [f"{first} and {second}"
for first , second in zip(first_month_names,second_month_names)]

```

**Task 7**

```python
import csv
fieldnames = [
    "county",
    "annual total mm",
    "wettest month",
    "dry months",
    "planting window"
]
with open("rainfall_summary.csv","w",newline="") as file:
  writer = csv.DictWriter(file,fieldnames=fieldnames)
  writer.writeheader()
  for i in range(len(highest_combinations)):
    writer.writerow({
        "county":counties[i],
        "annual total mm":annual_total_rain[i],
        "wettest month":wettest_month_names[i],
        "dry months":counts[i],
        "planting window": planting_windows[i]

    })





```

**Advisory For The NGO**           

***1.Farmers in Nairobi,Nakuru,Kisumu and Mombasa are advised to to prepare their farms ahead of the April-May rainfall period ,which recorded the highest combined rainfall in these counties.***

***2.Farmers in Machakos should prepare for November-December rainfall period instead which recorded the highest rainfall in the county.***

***3.Kisumu recorded the highest annual rainfall of 1,371mm,making proper drainage important to minimize waterlogging.***

***4.Machakos recorded the lowest annual rainfall of 811mm and six dry months,highlighting the importance of water conservation and drought preparedness.***
